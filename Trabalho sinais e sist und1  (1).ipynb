# Questão 1

Considere o sinal de tempo contínuo definido pela expressão:

$$
x(t) = (t^2 - 1)[u(t-1)-u(t-2)]
$$

(a) Plote o sinal $x(t)$ para $0 \leq t \leq 3$.

(b) Calcule, de forma analítica, a energia do sinal $x(t)$.

(c) Determine, de forma numérica, a energia do sinal $x(t)$. Compare com o resultado do item anterior.


## (a) Gráfico do sinal

```python
import numpy as np
import matplotlib.pyplot as plt

# Função degrau
def u(t):
    return np.where(t >= 0, 1, 0)

# Sinal x(t)
def x(t):
    return (t**2 - 1) * (u(t-1) - u(t-2))

# Tempo
t = np.linspace(0, 3, 1000)

# Plot
plt.figure(figsize=(10,3))
plt.plot(t, x(t))
plt.title("Sinal x(t)")
plt.xlabel("t")
plt.ylabel("x(t)")
plt.grid()
plt.show()
```

## (b) Energia analítica

A energia do sinal é dada por:

$$
E = \int_{-\infty}^{\infty} |x(t)|^2 dt
$$

Como o sinal existe apenas entre $1 \leq t < 2$:

$$
E = \int_1^2 (t^2 -1)^2 dt
$$

Expandindo:

$$
(t^2 -1)^2 = t^4 -2t^2 +1
$$

Então:

$$
E = \int_1^2 (t^4 -2t^2 +1)dt
$$

Integrando:

$$
E = \left[\frac{t^5}{5} - \frac{2t^3}{3} + t \right]_1^2
$$

Substituindo os limites:

$$
E =
\left(
\frac{32}{5}
-\frac{16}{3}
+2
\right)
-
\left(
\frac{1}{5}
-\frac{2}{3}
+1
\right)
$$

Resultado:

$$
E = \frac{38}{15} \approx 2.5333
$$

## (c) Energia numérica

```python
energia_numerica = np.trapezoid(x(t)**2, t)

print("Energia numérica:", energia_numerica)
```

## Comparação

Valor analítico:

$$
E = \frac{38}{15} \approx 2.5333
$$

Valor numérico obtido no Python:

$$
E \approx 2.519
$$

Os resultados são equivalentes.

## Questão 3

Um modulador em amplitude comercial é definido pela seguinte relação de entrada $x(t)$ e saída $y(t)$:

$$
y(t) = (A + x(t))\cos(2\pi f_c t)
$$

Determine, de forma computacional, a saída para $0 \leq t \leq 0,1$ do sistema modulador em amplitude quando a entrada é igual a:

$$
x(t) = \cos(2\pi f_m t)
$$

sendo $f_m = 50\ Hz$.

Considere $A = 1,5$ e $f_c = 500\ Hz$.

Plote a forma de onda do sinal de saída.

```python
import numpy as np
import matplotlib.pyplot as plt

# Parâmetros
fm = 50      # frequência da mensagem
fc = 500     # frequência da portadora
A = 1.5      # constante

# Tempo
t = np.linspace(0, 0.1, 10000)

# Sinal de entrada (mensagem)
x_t = np.cos(2 * np.pi * fm * t)

# Sinal de saída (modulado)
y_t = (A + x_t) * np.cos(2 * np.pi * fc * t)

# Plot
plt.figure(figsize=(10,4))
plt.plot(t, y_t)
plt.title("Sinal Modulado AM")
plt.xlabel("Tempo (s)")
plt.ylabel("y(t)")
plt.grid()
plt.show()
```

## Questão 4

Um determinado sistema linear e invariante no tempo possui uma resposta ao impulso $h(t)$ dada por:

$$
h(t) = e^{-2t}u(t)
$$

(a) Plote a resposta ao impulso do sistema para $0 \leq t \leq 3$.

(b) O sistema é estável? O sistema é causal? Justifique suas respostas.

(c) Considere que o seguinte sinal:

$$
x(t) = u(t) - u(t-2)
$$

é aplicado na entrada do sistema.

Determine, de forma analítica, a expressão da saída $y(t)$ para todo $t$ utilizando o método gráfico da convolução.

(d) Resolva o item (c) de forma numérica. Compare o resultado numérico com o resultado analítico, obtido no item (c). Uma forma de comparar consiste em plotar, em um mesmo gráfico, o sinal da saída obtido da forma analítica e o sinal de saída da forma numérica.

(e) Repetir os itens (a)-(d) quando a resposta ao impulso é igual a:

$$
h(t) = e^{t}u(t)
$$

## (a) Resposta ao impulso

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 3, 1000)
h = np.exp(-2*t)

plt.plot(t, h)
plt.title("Resposta ao impulso h(t)")
plt.xlabel("t")
plt.ylabel("h(t)")
plt.grid()
plt.show()
```

## (b) Estabilidade e causalidade

O sistema é causal, pois:

$$
h(t)=0 \quad para \quad t<0
$$

Logo, a resposta depende apenas de valores presentes e passados da entrada.

O sistema também é estável, pois a integral da resposta ao impulso é finita:

$$
\int_{-\infty}^{\infty}|h(t)|dt < \infty
$$

Como:

$$
h(t)=e^{-2t}u(t)
$$

a exponencial decai com o tempo, garantindo convergência da integral.

## (c) Saída analítica do sistema

O sinal de entrada é:

$$
x(t)=u(t)-u(t-2)
$$

e a resposta ao impulso é:

$$
h(t)=e^{-2t}u(t)
$$

A saída do sistema é dada pela convolução:

$$
y(t)=x(t)*h(t)
$$

Logo:

$$
y(t)=\int_{-\infty}^{\infty}x(\tau)h(t-\tau)d\tau
$$

Resolvendo a convolução, obtém-se:

$$
y(t)=
\frac{1}{2}(1-e^{-2t})u(t)
-
\frac{1}{2}(1-e^{-2(t-2)})u(t-2)
$$

## (d) Convolução numérica

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(-1, 5, 1000)
dt = t[1] - t[0]

# x(t)
x = np.where((t >= 0) & (t <= 2), 1, 0)

# h(t)
h = np.where(t >= 0, np.exp(-2*t), 0)

# convolução
y_num = np.convolve(x, h) * dt

# novo eixo de tempo
t_conv = np.linspace(t[0]+t[0], t[-1]+t[-1], len(y_num))

plt.plot(t_conv, y_num, label="Numérico")
plt.legend()
plt.grid()
plt.show()
```

## (d) Comparação Numerica e Analitica

```python
import numpy as np
import matplotlib.pyplot as plt

# tempo
t = np.linspace(-1, 5, 1000)
dt = t[1] - t[0]

# x(t)
x = np.where((t >= 0) & (t <= 2), 1, 0)

# h(t)
h = np.where(t >= 0, np.exp(-2*t), 0)

# convolução numérica
y_num = np.convolve(x, h) * dt

# eixo de tempo da convolução
t_conv = np.linspace(t[0]+t[0], t[-1]+t[-1], len(y_num))

# solução analítica
y_analitico = []

for ti in t_conv:
    if ti < 0:
        y_analitico.append(0)
    elif ti < 2:
        y_analitico.append((1 - np.exp(-2*ti))/2)
    else:
        y_analitico.append((np.exp(-2*ti)*(np.exp(4)-1))/2)

y_analitico = np.array(y_analitico)

# gráfico
plt.figure(figsize=(10,4))

plt.plot(t_conv, y_num, label="Numérico")
plt.plot(t_conv, y_analitico, '--', label="Analítico")

plt.title("Comparação entre solução analítica e numérica")
plt.xlabel("t")
plt.ylabel("y(t)")

plt.grid()
plt.legend()

plt.show()
```

Os resultados analítico e numérico são praticamente iguais, validando a convolução calculada.

## (e) Novo sistema

Para:

$$
h(t)=e^tu(t)
$$

o sistema deixa de ser estável, pois:

$$
\int_{-\infty}^{\infty}|h(t)|dt
$$

diverge.

Isso ocorre porque a exponencial cresce com o tempo.

```python
# 1. Configuração do Tempo e Sinais
t = np.linspace(-1, 5, 1000)
dt = t[1] - t[0]

# Entrada x(t) e Nova Resposta ao Impulso h(t)
x = np.where((t >= 0) & (t <= 2), 1, 0)
h = np.where(t >= 0, np.exp(t), 0)

# 2. Convolução Numérica
y_num = np.convolve(x, h) * dt
t_conv = np.linspace(2*t[0], 2*t[-1], len(y_num))

# 3. Solução Analítica (Item C)
y_analitico = np.piecewise(t_conv, 
    [(t_conv < 0), (t_conv >= 0) & (t_conv <= 2), (t_conv > 2)],
    [0, lambda t: np.exp(t) - 1, lambda t: np.exp(t) - np.exp(t - 2)]
)

# 4. Plotagem dos Resultados (Dois gráficos em uma figura)
plt.figure(figsize=(14, 5))

# Subplot 1: Resposta ao Impulso (Itens A e B)
plt.subplot(1, 2, 1)
plt.plot(t, h, color='green', label='$h(t) = e^t u(t)$', linewidth=2)
plt.fill_between(t, h, alpha=0.2, color='green')
plt.title("4(e) - Resposta ao Impulso (Sistema Instável)")
plt.xlabel("t")
plt.ylabel("h(t)")
plt.grid(True, linestyle=':')
plt.legend()

# Subplot 2: Comparação de Saídas (Item D)
plt.subplot(1, 2, 2)
plt.plot(t_conv, y_num, 'b', label='Convolução Numérica', linewidth=2.5)
plt.plot(t_conv, y_analitico, 'r--', label='Solução Analítica', linewidth=2)
plt.title("4(e) - Comparação: Saída Numérica vs Analítica")
plt.xlabel("t")
plt.ylabel("y(t)")
plt.xlim([-1, 5]) 
plt.grid(True, linestyle=':')
plt.legend()

plt.tight_layout()
plt.show()

# Justificativa teórica para o Item B
print("Análise Teórica:")
print("Causalidade: Causal (h(t)=0 para t<0).")
print("Estabilidade: Instável (Integral do módulo de h(t) de 0 a infinito diverge).")
```

## Questão 5

Um determinado circuito elétrico com tensão de entrada $v_i(t)$ e tensão de saída $v_o(t)$ é modelado através da seguinte equação diferencial ordinária:

$$
\frac{d^2v_o(t)}{dt^2}
+
7\frac{dv_o(t)}{dt}
+
10v_o(t)
=
\frac{dv_i(t)}{dt}
+
6v_i(t)
$$

As condições iniciais para esse circuito são:

$$
v_o(0)=6\ V
$$

e

$$
\frac{dv_o(0)}{dt} = -4\ V/s
$$

e a entrada é:

$$
x(t)=6e^{-3t}u(t)
$$

(a) Utilizando o método clássico, determine de forma analítica a saída do sistema $y(t)$ para todo $t$.

(b) Plote o sinal de saída para $0 \leq t \leq 2$.

(c) Utilizando computação simbólica, determine a expressão de saída para esse circuito quando a entrada é a mesma apresentada no item (a). Verifique e compare seu resultado com aquele obtido no item (a).

## (a) Solução analítica

A equação diferencial do sistema é:

$$
\frac{d^2y(t)}{dt^2}
+
7\frac{dy(t)}{dt}
+
10y(t)
=
\frac{dx(t)}{dt}
+
6x(t)
$$

A entrada é:

$$
x(t)=6e^{-3t}u(t)
$$

com condições iniciais:

$$
y(0)=6
$$

$$
y'(0)=-4
$$

A solução analítica encontrada é:

$$
y(t)=
\frac{1}{3}e^{-5t}
+
\frac{44}{3}e^{-2t}
-
9e^{-3t}
$$

## (b) Gráfico da saída

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 2, 1000)

vo = (1/3)*np.exp(-5*t) + (44/3)*np.exp(-2*t) - 9*np.exp(-3*t)

plt.plot(t, vo)
plt.title("Saída v_o(t)")
plt.xlabel("t")
plt.ylabel("v_o(t)")
plt.grid()
plt.show()
```

## (c) Computação simbólica

A solução simbólica foi utilizada para verificar a solução analítica encontrada anteriormente.

O resultado obtido foi:

$$
y(t)=
\frac{1}{3}e^{-5t}
+
\frac{44}{3}e^{-2t}
-
9e^{-3t}
$$

Assim, a solução simbólica confirmou a solução analítica.

```python
import sympy as sp
from IPython.display import display

t = sp.symbols('t')

vo = sp.Function('vo')

vi = 6*sp.exp(-3*t)

ode = sp.Eq(
    vo(t).diff(t,2) + 7*vo(t).diff(t) + 10*vo(t),
    sp.diff(vi,t) + 6*vi
)

ics = {
    vo(0): 6,
    sp.diff(vo(t), t).subs(t,0): -4
}

sol = sp.dsolve(ode, ics=ics)

display(sol)

display(
    sp.Eq(
        sp.Symbol('y(t)'),
        sp.simplify(sol.rhs)
    )
)
```
